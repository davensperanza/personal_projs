There are 5 jars, each containing 10 balls. You randomly choosae a jar and remove a ball. You repeat this until one of the jars is empty. On average, how many balls will you remove?

```python
import random
import matplotlib.pyplot as plt
```

```python
counts = []

for _ in range(100000):
    count = 0
    flag = True
    jars = [10,10,10,10,10] # reset

    while flag:
        jar_num = random.randint(0,4)
        jars[jar_num] -= 1
        if jars[jar_num] == 0:
            flag = False
        count += 1
    
    counts.append(count)
```

```python
# dict construction for bar graph

bars = {}
for i in range(1,51):
    bars[i] = 0

for el in counts:
    bars[el] += 1
```

```python
plt.title("Number of ball extracted distribution")
plt.bar(bars.keys(), bars.values())
plt.show()
```
